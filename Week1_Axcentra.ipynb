```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#Step 1: Load the dataset
# Make sure file is in same folder or give correct path
df = pd.read_csv('StudentsPerformance.csv')

print(" Dataset Loaded Successfully!\n")

#Step 2: Quick overview of data
print(" First 5 Rows of Dataset:")
print(df.head())

print("\n Dataset Info:")
print(df.info())

print("\n Missing Values:")
print(df.isnull().sum())


#Step 3: Data Cleaning
# Fill missing values (if any)
df.fillna(df.mean(numeric_only=True), inplace=True)

print("\nData Cleaning Done!\n")


#Step 4: Basic Statistics
print("Basic Statistics of Scores:\n")

print("Average Scores:")
print(df[['math score','reading score','writing score']].mean())

print("\nMaximum Scores:")
print(df[['math score','reading score','writing score']].max())

print("\nMinimum Scores:")
print(df[['math score','reading score','writing score']].min())


# Step 5: Group-wise Analysis

# Gender-wise performance
print("\n Gender-wise Average Scores:")
gender_avg = df.groupby('gender')[['math score','reading score','writing score']].mean()
print(gender_avg)

# Test preparation impact
print("\n Test Preparation Impact:")
test_prep = df.groupby('test preparation course')[['math score','reading score','writing score']].mean()
print(test_prep)


# Step 6: Visualization (Graphs)

# Graph 1: Math score by gender
plt.figure()
df.groupby('gender')['math score'].mean().plot(kind='bar')
plt.title("Average Math Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Math Score")
plt.show()


# Graph 2: Test preparation impact
plt.figure()
df.groupby('test preparation course')['math score'].mean().plot(kind='bar')
plt.title("Impact of Test Preparation on Math Score")
plt.xlabel("Test Preparation")
plt.ylabel("Math Score")
plt.show()


#Step 7: Final Conclusion
print("\n Conclusion:")
print("1. Students who completed the test preparation course scored higher.")
print("2. Female students performed better in reading and writing.")
print("3. Math scores are slightly lower compared to reading and writing.")
print("4. Data cleaning ensured no missing values affected the analysis.")


```
